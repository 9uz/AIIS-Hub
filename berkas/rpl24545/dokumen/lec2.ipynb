### Memilih dan Mengambil Data dari `DataFrame`

Terdapat tiga metode utama untuk memilih data pada `pandas`: `iloc[]`, `loc[]`, dan seleksi berbasis konteks.

#### **1. `iloc[]` - Pengindeksan Berbasis Posisi Integer**
- Digunakan jika Anda ingin memilih data berdasarkan posisi urutan baris dan kolom (dimulai dari angka 0).
- Contoh:
    - Memilih sel spesifik: `df.iloc[0, 1]` (baris ke-1, kolom ke-2).
    - Memilih rentang baris dan kolom: `df.iloc[1:3, 0:2]`.

#### **2. `loc[]` - Pengindeksan Berbasis Label**
- Digunakan jika Anda ingin memilih data berdasarkan label nama baris dan nama kolom.
- Contoh:
    - Memilih sel spesifik: `df.loc[0, 'Age']` (baris berindeks `0`, kolom bernama `'Age'`).
    - Memilih rentang label: `df.loc[1:3, 'Name':'City']`.

#### **3. Seleksi Berbasis Konteks**
- Memungkinkan Anda memilih data secara langsung menggunakan nama kolom atau irisan baris.
- Contoh:
    - Memilih satu kolom: `df['Name']`.
    - Memilih beberapa kolom: `df[['Name', 'Age']]`.
    - Memilih potongan baris: `df[1:3]`.

```python
# Mengakses elemen pada baris pertama, kolom kedua dengan iloc
display(df.iloc[0, 1])
```

```python
# Mengakses seluruh baris pada kolom kedua dengan iloc
display(df.iloc[:, 1])
```

```python
# Mengakses seluruh kolom pada baris pertama dengan iloc
display(df.iloc[0])
```

```python
# Mengakses potongan baris 1 sampai 2 (indeks 3 tidak termasuk) dengan iloc
display(df.iloc[1:3])
```

```python
# Mengakses potongan kolom 1 sampai 2 untuk semua baris dengan iloc
display(df.iloc[:, 1:3])
```

```python
# Mengakses kombinasi rentang baris dan kolom tertentu dengan iloc
display(df.iloc[1:3, 1:3])
```

```python
# Mengakses elemen pada baris indeks 0 dan kolom 'Landmark' dengan loc
display(df.loc[0, 'Landmark'])
```

```python
# Mengakses seluruh baris untuk kolom 'Landmark' dengan loc
display(df.loc[:, 'Landmark'])
```

```python
# Mengakses seluruh kolom untuk baris indeks 0 dengan loc
display(df.loc[0])
```

### Memfilter Data pada `DataFrame`

Pemfilteran (*filtering*) memungkinkan kita mengambil baris-baris data yang hanya memenuhi kriteria tertentu.

Proses pemfilteran dapat digabungkan dengan operator logika seperti `&` (DAN/AND), `|` (ATAU/OR), dan `~` (BUKAN/NOT) untuk query kondisi yang lebih kompleks.

```python
# Filter berdasarkan satu kondisi (Height > 50)
display(df[df['Height'] > 50])
```

```python
# Filter berdasarkan beberapa kondisi sekaligus (Height > 50 DAN Type == 'Library')
display(df[(df['Height'] > 50) & (df['Type'] == 'Library')])
```

```python
# Filter menggunakan isin() untuk mencocokkan beberapa nilai kategori
display(df[df['Type'].isin(['Gate', 'Plaza'])])
```

### Memodifikasi `DataFrame`

Kita dapat mengubah struktur dan isi `DataFrame` dengan:
1. Menambahkan kolom baru.
2. Melakukan kalkulasi aritmatika untuk membuat kolom baru.
3. Mengubah nilai data yang sudah ada.
4. Menghapus kolom yang tidak diperlukan.

```python
# Menambahkan kolom baru bernama 'Experience'
df['Experience'] = [2, 5, 1, 8, 4]
display(df)
```

```python
# Menambahkan kolom kalkulasi (contoh: estimasi kenaikan tinggi 10%)
df['Height_Increase'] = df['Height'] * 0.10
display(df)
```

```python
# Mengubah nilai data spesifik menggunakan .loc[]
df.loc[df['Landmark'] == 'Sather Gate', 'Height_Increase'] = 5
display(df)
```

```python
# Menghapus kolom 'Experience' dari DataFrame
# Catatan: Operasi ini tidak langsung mengubah DataFrame asli karena inplace=False secara default
df.drop(columns=['Experience'])
display(df)
```

```python
# Menghapus kolom 'Experience' secara permanen dengan parameter inplace=True
df.drop(columns=['Experience'], inplace=True)

display(df)
```

```python
# Menugaskan kembali hasil drop ke variabel baru
df_dropped = df.drop(columns=['Height_Increase'])

display(df_dropped)

# Menampilkan df asli untuk membuktikan bahwa data awal belum berubah sebelum ditimpa
display(df)
```

```python
# Menimpa variabel df dengan DataFrame baru yang sudah dibersihkan
df = df_dropped

df
```

#### Menangani Nilai yang Hilang (*Missing Values*)

Data kosong atau *missing value* adalah masalah umum pada dataset dunia nyata. Penanganan yang tepat sangat penting untuk menjaga keakuratan analisis.

Kita akan mempraktikkan teknik untuk:
1. Mendeteksi nilai yang hilang.
2. Menghitung jumlah nilai yang hilang.
3. Menangani nilai kosong dengan cara menghapus (*dropping*) atau mengisinya (*imputation*).

```python
# Simulasi pembuatan data kosong (NaN / None)
df_missing = df.copy()
df_missing.loc[0, 'Year Built'] = np.nan
df_missing.loc[2, 'Height'] = np.nan
df_missing.loc[4, 'Type'] = None
display(df_missing)
```

```python
# Memeriksa lokasi data yang kosong (menghasilkan boolean True/False)
display(df_missing.isnull())

# Menghitung jumlah nilai kosong per kolom
display(df_missing.isnull().sum())
```

```python
# Menghapus baris yang mengandung nilai kosong
display(df_missing.dropna())
```

```python
# Mengisi seluruh nilai kosong dengan angka 0
display(df_missing.fillna(0))
```

```python
# Mengisi nilai kosong sesuai karakteristik kolom masing-masing:
# Rata-rata untuk 'Year Built', Median untuk 'Height', dan 'Unknown' untuk 'Type'
display(df_missing.fillna({
    'Year Built': df_missing['Year Built'].mean(),
    'Height': df_missing['Height'].median(),
    'Type': 'Unknown'
}))
```

### Agregasi pada `DataFrame`

Agregasi digunakan untuk meringkas data, seperti menghitung rata-rata, total jumlah, dan statistik ringkasan lainnya.

1. **Fungsi Agregasi**: Fungsi seperti `mean()`, `sum()`, `count()`, dan sejenisnya dapat diterapkan langsung pada kolom.
2. **Pengelompokan (*Grouping*)**: Mengelompokkan data berdasarkan kriteria tertentu sebelum menerapkan fungsi agregasi.

    - **Sintaks**: Gunakan `.groupby()` diikuti oleh fungsi agregasi.
    - **Contoh**:
      - Rata-rata per kategori: `df.groupby('Type')['Height'].mean()`
      - Total berdasarkan beberapa kolom: `df.groupby(['Type', 'Year Built'])['Height'].sum()`
      - Menghitung jumlah entri per kelompok: `df.groupby('Type').size()`

#### Daftar Fungsi Agregasi

##### Agregasi Dasar
- `sum()` – jumlah total nilai
- `mean()` – nilai rata-rata
- `median()` – nilai tengah (*median*)
- `min()` – nilai minimum
- `max()` – nilai maksimum
- `count()` – jumlah entri tidak kosong
- `nunique()` – jumlah nilai unik
- `prod()` – hasil kali seluruh nilai

##### Agregasi Statistik
- `std()` – standar deviasi
- `var()` – varians
- `sem()` – *standard error of the mean*
- `skew()` – kemiringan distribusi (*skewness*)

##### Agregasi Logika & Berbasis Indeks
- `any()` – bernilai True jika ada minimal satu nilai True
- `all()` – bernilai True jika seluruh nilai bernilai True
- `first()` – mengambil nilai pertama yang tidak kosong
- `last()` – mengambil nilai terakhir yang tidak kosong
- `idxmin()` – mengambil posisi indeks dari nilai terkecil
- `idxmax()` – mengambil posisi indeks dari nilai terbesar

```python
# Menghitung rata-rata kolom 'Height'
height_mean = df['Height'].mean()
print("Rata-rata tinggi:", height_mean)

# Menghitung total jumlah kolom 'Height'
height_sum = df['Height'].sum()
print("Total tinggi:", height_sum)
```

```python
# Menghitung standar deviasi kolom 'Height'
height_std = df['Height'].std()
print("Standar deviasi tinggi:", height_std)
```

```python
# Mencari posisi indeks dari nilai tertinggi pada kolom 'Height'
max_height_index = df['Height'].idxmax()
print("Indeks dengan tinggi maksimum:", max_height_index)
```
