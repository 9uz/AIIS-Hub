## Persiapan

```python
# [KODE] Alat pemeriksa jawaban — jalankan sekali di awal
import hashlib
import numpy as np
import matplotlib.pyplot as plt

def periksa(label, jawaban_anda, sandi_benar, desimal=3):
    if jawaban_anda is None:
        print(f'  ○ {label:38s} belum diisi')
        return
    cap = hashlib.sha256(f'{round(float(jawaban_anda), desimal):.{desimal}f}'.encode()).hexdigest()[:16]
    if cap == sandi_benar:
        print(f'  ✓ {label:38s} cocok')
    else:
        print(f'  ✗ {label:38s} belum cocok — periksa kembali langkah Anda')

print('Alat pemeriksa siap. Isi jawaban Anda pada sel [PERIKSA], lalu jalankan.')
```

```python
# [KODE] Memeriksa penelusuran Anda secara visual
import matplotlib.pyplot as plt

titik = np.array([[1,1],[1,2],[2,1],[6,5],[7,5],[6,6]], dtype=float)
label_titik = ['A','B','C','D','E','F']
sentroid = np.array([[0.5,0.5],[3.0,0.5]])

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6), sharex=True, sharey=True)
warna = ['#C0504D', '#2E75B6']

for langkah in range(3):
    ax = axes[langkah]
    jarak = np.linalg.norm(titik[:, None, :] - sentroid[None, :, :], axis=2)
    anggota = jarak.argmin(axis=1)
    for i, (t, lab) in enumerate(zip(titik, label_titik)):
        w = warna[anggota[i]] if langkah > 0 else '#9AA5B1'
        ax.scatter(*t, s=90, color=w, zorder=3)
        ax.annotate(lab, (t[0]+0.18, t[1]+0.12), fontsize=9)
    for j, c in enumerate(sentroid):
        ax.scatter(*c, marker='X', s=180, color=warna[j], edgecolor='white', lw=1.5, zorder=4)
    ax.set_title(['Awal', 'Iterasi 1', 'Iterasi 2'][langkah], fontsize=11)
    ax.grid(alpha=0.3); ax.set_xlim(-0.5, 8); ax.set_ylim(-0.5, 7.5)
    if langkah < 2:
        sentroid = np.array([titik[anggota == j].mean(axis=0) for j in range(2)])

plt.tight_layout(); plt.show()
print('Sentroid akhir:'); print(np.round(sentroid, 3))
```

Sekarang bayangkan ciri kedua **tidak diskalakan**: tingkat keterlambatan diukur dalam menit, dengan nilai puluhan kali lebih besar.

```python
# [KODE] Membandingkan hasil dengan dan tanpa penskalaan
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Ciri kedua dikalikan 50: seolah diukur dalam satuan yang jauh lebih besar
titik_asli   = np.array([[1,1],[1,2],[2,1],[6,5],[7,5],[6,6]], dtype=float)
titik_timpang = titik_asli.copy()
titik_timpang[:, 1] *= 50

hasil_timpang = KMeans(n_clusters=2, n_init=10, random_state=0).fit_predict(titik_timpang)
hasil_skala   = KMeans(n_clusters=2, n_init=10, random_state=0).fit_predict(
                    StandardScaler().fit_transform(titik_timpang))

print('Titik           :', label_titik)
print('Tanpa penskalaan:', hasil_timpang)
print('Dengan penskalaan:', hasil_skala)
```
