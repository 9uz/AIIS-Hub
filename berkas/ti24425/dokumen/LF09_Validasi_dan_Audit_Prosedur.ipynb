## Persiapan

```python
# [KODE] Alat pemeriksa jawaban — jalankan sekali di awal
import hashlib
import numpy as np
import matplotlib.pyplot as plt

def periksa(label, jawaban_anda, sandi_benar, desimal=3):
    if jawaban_anda is None:
        print(f'  ○ {label:38s} belum diisi')
        return
    cap = hashlib.sha256(f'{round(float(jawaban_anda), desimal):.{desimal}f}'.encode()).hexdigest()[:16]
    if cap == sandi_benar:
        print(f'  ✓ {label:38s} cocok')
    else:
        print(f'  ✗ {label:38s} belum cocok — periksa kembali langkah Anda')

print('Alat pemeriksa siap. Isi jawaban Anda pada sel [PERIKSA], lalu jalankan.')
```

# Soal 4 &mdash; Membuktikan Sendiri Akibat Kebocoran

Sel berikut menjalankan dua prosedur pada data yang sama: satu bocor, satu sahih. **Perkirakan dulu** hasil mana yang akan lebih tinggi sebelum menjalankannya.

```python
# [KODE] Membandingkan prosedur bocor dan prosedur sahih
from sklearn.datasets import make_classification
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import cross_val_score, train_test_split, KFold
from sklearn.pipeline import Pipeline

# Data buatan: 300 baris, 500 fitur, TIDAK ADA hubungan nyata dengan label
X, y = make_classification(n_samples=300, n_features=500, n_informative=2,
                           n_redundant=0, shuffle=False, random_state=42)
y = np.random.RandomState(0).permutation(y)     # label diacak: hubungan dihapus total

# --- Prosedur BOCOR: seleksi fitur dilakukan pada SELURUH data lebih dulu ---
X_pilih = SelectKBest(f_classif, k=20).fit_transform(X, y)
skor_bocor = cross_val_score(LogisticRegression(max_iter=2000), X_pilih, y,
                             cv=KFold(5, shuffle=True, random_state=1))

# --- Prosedur SAHIH: seleksi fitur berada DI DALAM Pipeline ---
pipa = Pipeline([('pilih', SelectKBest(f_classif, k=20)),
                 ('model', LogisticRegression(max_iter=2000))])
skor_sahih = cross_val_score(pipa, X, y, cv=KFold(5, shuffle=True, random_state=1))

print('Label sudah diacak, jadi akurasi yang jujur seharusnya sekitar 0,50\n')
print(f'Prosedur BOCOR  : {skor_bocor.mean():.3f}')
print(f'Prosedur SAHIH  : {skor_sahih.mean():.3f}')
```

**Pertanyaan penalaran 4.1** Label pada percobaan di atas sudah **diacak sepenuhnya**, sehingga tidak ada hubungan apa pun antara fitur dan label. Akurasi yang jujur seharusnya sekitar 0,50.
