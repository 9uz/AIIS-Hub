## C.3 Rancangan Pembagian Data

Sebelum menjalankan sel di bawah, tentukan lebih dulu skema pembagian yang **tepat** untuk dataset Anda. Rujuk pembahasan pertemuan 3 tentang tiga situasi khusus: data tidak seimbang, data deret waktu, dan data berkelompok.

| Butir | Isian |
|---|---|
| Skema yang dipilih (acak / stratified / berdasarkan waktu / group) | |
| Alasan pemilihan skema | |
| Proporsi latih : validasi : uji | |
| Kolom penentu pembagian (bila stratified atau group) | |

```python
# [KODE] Pembagian data — sesuaikan dengan skema yang Anda pilih
from sklearn.model_selection import train_test_split

KOLOM_TARGET = 'ganti_dengan_nama_kolom_target'

X = df.drop(columns=[KOLOM_TARGET])
y = df[KOLOM_TARGET]

# ---- Pilih SALAH SATU skema di bawah ini, beri tanda pagar pada yang tidak dipakai ----

# Skema 1 — acak biasa
X_sisa, X_uji, y_sisa, y_uji = train_test_split(
    X, y, test_size=0.20, random_state=42)

# Skema 2 — stratified (untuk klasifikasi dengan kelas tidak seimbang)
# X_sisa, X_uji, y_sisa, y_uji = train_test_split(
#     X, y, test_size=0.20, random_state=42, stratify=y)

# Skema 3 — berdasarkan waktu (data deret waktu; JANGAN diacak)
# df_urut = df.sort_values('ganti_dengan_kolom_tanggal')
# batas   = int(len(df_urut) * 0.80)
# latih_val, uji = df_urut.iloc[:batas], df_urut.iloc[batas:]

# ---- Pisahkan sisanya menjadi latih dan validasi ----
X_latih, X_val, y_latih, y_val = train_test_split(
    X_sisa, y_sisa, test_size=0.25, random_state=42)

print('Jumlah baris data latih    :', len(X_latih))
print('Jumlah baris data validasi :', len(X_val))
print('Jumlah baris data uji      :', len(X_uji))
print()
print('Proporsi : {:.0f} : {:.0f} : {:.0f}'.format(
    len(X_latih)/len(df)*100, len(X_val)/len(df)*100, len(X_uji)/len(df)*100))
```

```python
# [KODE] Periksa apakah proporsi kelas tetap terjaga (khusus masalah klasifikasi)
# Lewati sel ini bila masalah Anda berupa regresi atau clustering.
try:
    banding = pd.DataFrame({
        'seluruh_data' : y.value_counts(normalize=True).round(3),
        'data_latih'   : y_latih.value_counts(normalize=True).round(3),
        'data_uji'     : y_uji.value_counts(normalize=True).round(3),
    })
    print(banding)
except Exception as e:
    print('Sel ini dilewati:', e)
```
